*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def game_status(board):
    n = len(board)

    # Check rows
    for row in board:
        if row[0] != 0 and all(cell == row[0] for cell in row):
            return row[0]

    # Check columns
    for col in range(n):
        if board[0][col] != 0 and all(board[row][col] == board[0][col] for row in range(n)):
            return board[0][col]

    # Check main diagonal
    if board[0][0] != 0 and all(board[i][i] == board[0][0] for i in range(n)):
        return board[0][0]

    # Check other diagonal
    if board[0][n - 1] != 0 and all(board[i][n - 1 - i] == board[0][n - 1] for i in range(n)):
        return board[0][n - 1]

    # If there are empty spaces, the game is incomplete
    if any(0 in row for row in board):
        return -1

    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def make_move(board, player, row, col):
    if board[row][col] != 0:
        return False

    if player not in (1, 2):
        return False

    board[row][col] = player
    return True
````

*Exercise 6:* Modify Exercise 3 to show **row and column labels** so that players can specify locations like `"A2"` or `"C1"`.

````python
def draw_labeled_board(board):
    n = len(board)

    # Column labels
    print("    " + "   ".join(str(i + 1) for i in range(n)))

    for row_index, row in enumerate(board):
        print("   " + "--- ".join(["---"] * n))
        values = []
        for cell in row:
            if cell == 1:
                values.append(" X ")
            elif cell == 2:
                values.append(" O ")
            else:
                values.append("   ")
        print(chr(65 + row_index) + " |" + "|".join(values) + "|")

    print("   " + "--- ".join(["---"] * n))
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
def make_location_move(board, player, location):
    location = location.strip().upper()

    if len(location) < 2:
        return False

    row_letter = location[0]
    col_text = location[1:]

    if not row_letter.isalpha() or not col_text.isdigit():
        return False

    row = ord(row_letter) - ord("A")
    col = int(col_text) - 1

    if row < 0 or row >= len(board) or col < 0 or col >= len(board):
        return False

    return make_move(board, player, row, col)
````

````python
test_board = create_board(3)
print(make_location_move(test_board, 1, "A2"))  # True
print(make_location_move(test_board, 2, "C1"))  # True
print(make_location_move(test_board, 1, "A2"))  # False
draw_labeled_board(test_board)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

````python
def player_move(board, player):
    while True:
        location = input("Player " + str(player) + ", enter a location (example: A2): ")

        if make_location_move(board, player, location):
            return True

        print("Invalid move. Please enter an empty location such as A2.")
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def play_game(n=3):
    board = create_board(n)
    player = 1

    while True:
        draw_labeled_board(board)
        player_move(board, player)

        status = game_status(board)

        if status == 1:
            draw_labeled_board(board)
            print("Player 1 (X) wins!")
            break
        elif status == 2:
            draw_labeled_board(board)
            print("Player 2 (O) wins!")
            break
        elif status == 0:
            draw_labeled_board(board)
            print("The game is a draw!")
            break

        player = 2 if player == 1 else 1
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
import random

def computer_move_simple(board, computer, opponent):
    n = len(board)

    # First, look for a winning move.
    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = computer
                if game_status(board) == computer:
                    return True
                board[r][c] = 0

    # Next, block the opponent's winning move.
    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = opponent
                if game_status(board) == opponent:
                    board[r][c] = computer
                    return True
                board[r][c] = 0

    # Otherwise, take the center if it is available.
    if n % 2 == 1 and board[n // 2][n // 2] == 0:
        board[n // 2][n // 2] = computer
        return True

    # Otherwise, choose the first available square.
    empty = [(r, c) for r in range(n) for c in range(n) if board[r][c] == 0]
    if empty:
        r, c = random.choice(empty)
        board[r][c] = computer
        return True

    return False


def play_vs_computer(n=3):
    board = create_board(n)
    human = 1
    computer = 2
    player = human

    while True:
        draw_labeled_board(board)

        if player == human:
            player_move(board, human)
        else:
            computer_move_simple(board, computer, human)
            print("Computer moved.")

        status = game_status(board)

        if status != -1:
            draw_labeled_board(board)
            if status == 0:
                print("The game is a draw!")
            else:
                print("Player", status, "wins!")
            break

        player = computer if player == human else human
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def evaluate_board(board, computer, opponent):
    status = game_status(board)

    if status == computer:
        return 1
    if status == opponent:
        return -1
    return 0


def minimax(board, computer, opponent, depth):
    status = game_status(board)

    if status != -1:
        return evaluate_board(board, computer, opponent)

    if depth == 0:
        return 0

    best_score = -2

    for r in range(len(board)):
        for c in range(len(board)):
            if board[r][c] == 0:
                board[r][c] = computer
                score = -minimax(board, opponent, computer, depth - 1)
                board[r][c] = 0

                if score > best_score:
                    best_score = score

    return best_score


def computer_move_search(board, computer, opponent, max_depth=3):
    best_score = -2
    best_move = None

    for r in range(len(board)):
        for c in range(len(board)):
            if board[r][c] == 0:
                board[r][c] = computer
                score = -minimax(board, opponent, computer, max_depth - 1)
                board[r][c] = 0

                if score > best_score:
                    best_score = score
                    best_move = (r, c)

    if best_move is not None:
        r, c = best_move
        board[r][c] = computer
        return True

    return False


def play_vs_search_computer(n=3, max_depth=4):
    board = create_board(n)
    human = 1
    computer = 2
    player = human

    while True:
        draw_labeled_board(board)

        if player == human:
            player_move(board, human)
        else:
            computer_move_search(board, computer, human, max_depth)
            print("Computer moved.")

        status = game_status(board)

        if status != -1:
            draw_labeled_board(board)
            if status == 0:
                print("The game is a draw!")
            else:
                print("Player", status, "wins!")
            break

        player = computer if player == human else human
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def computer_vs_computer(n, depth1=2, depth2=1, games=10):
    smarter_wins = 0
    draws = 0
    weaker_wins = 0

    for _ in range(games):
        board = create_board(n)
        player = 1

        while game_status(board) == -1:
            if player == 1:
                computer_move_search(board, 1, 2, depth1)
            else:
                computer_move_search(board, 2, 1, depth2)

            player = 2 if player == 1 else 1

        status = game_status(board)

        if status == 1:
            smarter_wins += 1
        elif status == 2:
            weaker_wins += 1
        else:
            draws += 1

    return smarter_wins, weaker_wins, draws


def run_experiment():
    # Small depths are used so the experiments finish quickly.
    for n, depth1, depth2 in [(3, 3, 1), (4, 2, 1), (5, 1, 1)]:
        smarter, weaker, draws = computer_vs_computer(
            n, depth1=depth1, depth2=depth2, games=10
        )
        win_rate = smarter / 10
        print(
            str(n) + "x" + str(n),
            "smarter wins:", smarter,
            "weaker wins:", weaker,
            "draws:", draws,
            "smarter win rate:", win_rate
        )
````

````python
run_experiment()
````
